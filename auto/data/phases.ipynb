```python
import numpy as np
import pandas as pd
```

```python
# read in phases data
phases_df = pd.read_csv('phases_lyons.csv')
phases_df
```

```python
# get Channel col as list
channels = phases_df['Channel'].tolist()
split_channels = [chan.split(' ') for chan in channels]
networks = {chan[0] for chan in split_channels}
networks
```

```python
# get range of Distance values
distances = phases_df['Distance'].tolist()
max_distance = max(distances)
min_distance = min(distances)
print(f'Min Distance: {min_distance}, Max Distance: {max_distance}')
```
